```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
from IPython.core.pylabtools import figsize
# ESCENARIO 1 (phi3:mini)

# Estilo de los gráficos
sns.set_theme(style="whitegrid")

# Cargar los DataFrames
df_raw_c = pd.read_csv('../results/prompt1/escenario1_resultados_raw_completo_phi3mini.csv')
df_raw_r = pd.read_csv('../results/prompt1/escenario1_resultados_raw_reducido_phi3mini.csv')
df_json_r = pd.read_csv('../results/prompt1/escenario1_resultados_json_reducido_phi3mini.csv')

# Agregar una columna para identificar de dónde viene cada dato
df_raw_c['Formato'] = 'RAW Completo'
df_raw_r['Formato'] = 'RAW Reducido'
df_json_r['Formato'] = 'JSON Reducido'

# Unir todo en un único DataFrame
df_total = pd.concat([df_raw_c, df_raw_r, df_json_r], ignore_index=True)

# Asegurar el orden de los riesgos
orden_riesgo = ['INFO', 'BAJO', 'MEDIO', 'ALTO', 'INFORMACIÓN', 'ERROR_PARSE']

# ==========================================
# Gráfico 1: Distribución de Riesgos
# ==========================================
plt.figure(figsize=(10, 6))
# countplot cuenta automáticamente las ocurrencias en la columna Riesgo
ax = sns.countplot(
    data=df_total, 
    x='Formato', 
    hue='Riesgo', 
    hue_order=orden_riesgo,
    palette=['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728', 'purple', 'yellow']
)

for container in ax.containers:
    ax.bar_label(container, fontsize=10, padding=3)

plt.title('Distribución de Clasificación de Riesgo por Formato (Escenario 1 - phi3:mini)', fontsize=14, pad=15)
plt.xlabel('Formato de Entrada', fontsize=12)
plt.ylabel('Cantidad de Eventos', fontsize=12)
plt.legend(title='Nivel de Riesgo', bbox_to_anchor=(1.05, 1), loc='upper left')

# Guardar la imagen
#plt.savefig('../results/charts/grafico_distribucion_escenario1.png', dpi=300, bbox_inches='tight')
plt.show()

# ==========================================
# Gráfico 2: Tiempo Promedio de Respuesta
# ==========================================

# Promedio omitiendo los que tienen el tiempo en 0.0 porque el LLM no los procesó
mean_tr_rawr = df_raw_r[df_raw_r['Tiempo Respuesta (s)'] != 0.0]['Tiempo Respuesta (s)'].mean()
mean_tr_rawc = df_raw_c[df_raw_c['Tiempo Respuesta (s)'] != 0.0]['Tiempo Respuesta (s)'].mean()
mean_tr_jsonr = df_json_r[df_json_r['Tiempo Respuesta (s)'] != 0.0]['Tiempo Respuesta (s)'].mean()

valores_medias = [mean_tr_rawc, mean_tr_rawr, mean_tr_jsonr]

etiquetas = ['RAW Completo','RAW Reducido','JSON Reducido']

fig, ax = plt.subplots(figsize=(8, 6)) 

ax.bar(etiquetas, valores_medias, color=['blue', 'orange', 'green'])
ax.set_title('Tiempo Promedio de Respuesta por Formato (Escenario 1 - phi3:mini)')
ax.set_ylabel('Tiempo Promedio (s)')
ax.set_xlabel('Formato de Entrada')

for i, v in enumerate(valores_medias):
    ax.text(i, v + (max(valores_medias) * 0.01), f"{v:.2f}", ha='center', fontweight='bold')

# Guardar la imagen
plt.savefig('../results/charts/grafico_tiempos_escenario1.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# COMPARACIÓN DEL TIEMPO DE RESUPESTA DE LOS MODELOS

# Cargar los DataFrames
# Escenarios 1,2 y 5 con prompt1 en formato raw completo del modelo phi3:mini
df_e1_phi = pd.read_csv('../results/prompt1/escenario1_resultados_raw_completo_phi3mini.csv')
df_e2_phi = pd.read_csv('../results/prompt1/escenario2_resultados_raw_completo_phi3mini.csv')
df_e5_phi = pd.read_csv('../results/prompt1/escenario1_resultados_raw_completo_phi3mini.csv')

# Escenarios 1,2 y 5 con prompt1 en formato raw completo del modelo llama3.2:3b
df_e1_llama = pd.read_csv('../results/prompt1/escenario1_resultados_raw_completo_llama3-2_3b.csv')
df_e2_llama = pd.read_csv('../results/prompt1/escenario2_resultados_raw_completo_llama3-2_3b.csv')
df_e5_llama = pd.read_csv('../results/prompt1/escenario1_resultados_raw_completo_llama3-2_3b.csv')

mean_tr_e1_phi = df_e1_phi['Tiempo Respuesta (s)'].mean()
mean_tr_e1_llama = df_e1_llama['Tiempo Respuesta (s)'].mean()
mean_tr_e2_phi = df_e2_phi['Tiempo Respuesta (s)'].mean()
mean_tr_e2_llama = df_e2_llama['Tiempo Respuesta (s)'].mean()
mean_tr_e5_phi = df_e5_phi['Tiempo Respuesta (s)'].mean()
mean_tr_e5_llama = df_e5_llama['Tiempo Respuesta (s)'].mean()

# 1. Reorganizamos los datos por escenario y por modelo
escenarios = ['Escenario 1', 'Escenario 2', 'Escenario 5']
medias_phi = [mean_tr_e1_phi, mean_tr_e2_phi, mean_tr_e5_phi]
medias_llama = [mean_tr_e1_llama, mean_tr_e2_llama, mean_tr_e5_llama]

# 2. Configuramos las posiciones en el eje X y el ancho de las barras
x = np.arange(len(escenarios))  # Posiciones: [0, 1, 2]
ancho = 0.35  # Ancho de cada barra

fig, ax = plt.subplots(figsize=(10, 6))

# 3. Dibujamos las barras, desplazándolas hacia la izquierda y derecha
barras_phi = ax.bar(x - ancho/2, medias_phi, ancho, label='phi3:mini', color='blue')
barras_llama = ax.bar(x + ancho/2, medias_llama, ancho, label='llama3.2:3b', color='orange')

# 4. Ajustamos las etiquetas y el título
ax.set_title('Tiempo Promedio de Respuesta por Escenario - Modelo')
ax.set_ylabel('Tiempo Promedio (s)')
ax.set_xticks(x)
ax.set_xticklabels(escenarios)
ax.legend() # Esto agrega el cuadro con los colores y nombres de los modelos

# 5. Función para agregar los valores exactos encima de cada barra
def agregar_etiquetas(barras):
    for barra in barras:
        altura = barra.get_height()
        ax.annotate(f'{altura:.2f}',
                    xy=(barra.get_x() + barra.get_width() / 2, altura),
                    xytext=(0, 3),  # Desplazamiento vertical de 3 puntos
                    textcoords="offset points",
                    ha='center', va='bottom', fontweight='bold')

agregar_etiquetas(barras_phi)
agregar_etiquetas(barras_llama)

# Guardar y mostrar
plt.savefig('../results/charts/grafico_comparación_tiempos_modelos.png', dpi=300, bbox_inches='tight')
plt.show()
```
